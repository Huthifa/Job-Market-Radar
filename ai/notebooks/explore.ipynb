# Hacker News: "Who is hiring?"

```python
import requests

r = requests.get(
    "https://hn.algolia.com/api/v1/search_by_date",
    params={"tags": "story,author_whoishiring", "hitsPerPage": 5},
    timeout=30,
)
r.raise_for_status()

for hit in r.json()["hits"]:
    print(hit["objectID"], hit["title"])
```

```python
thread_id = "50338728"

thread = requests.get(f"https://hn.algolia.com/api/v1/items/{thread_id}", timeout=30).json()
comments = thread["children"]

print(len(comments))
print(comments[0].keys())
```

```python
for c in comments[:3]:
    print(c["text"][:500] if c["text"] else "<<EMPTY>>")
    print("-" * 60)
```

```python
import html

total = 0
with_pipes = 0

for c in comments:
    if c["text"]:
        text = html.unescape(c["text"])
        first_line = text.split("<p>")[0]
        total = total + 1
        if "|" in first_line:
            with_pipes = with_pipes + 1

print("Posts with text:", total)
print("Posts using the | pattern:", with_pipes)
```
